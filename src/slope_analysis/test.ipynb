```python
import numpy as np


samples = np.array([[1,4,3,5,8],
                    [3,2,4,9,8],
                    [1,1,1,3,7],
                    [1,10,1,3,10]])
weights = np.array([0.2, 0.1, 0.4,0.1,0.2])
xs = np.array([2, 5, 10])
axis = -1
print(samples.shape)
```

```python

ind_order = np.argsort(samples, axis)
print(ind_order)
```

```python
sorted_samples = np.take_along_axis(samples, ind_order, axis=axis)
print(sorted_samples)
```

```python
sorted_weights = weights[ind_order]
print(sorted_weights)
```

```python
cumsum_weights = np.insert(np.cumsum(sorted_weights, axis=axis), 0,0, axis=axis)
print(cumsum_weights)
# Find indices where sorted_samples < x
```

```python
o=np.vstack([cumsum_weights[i, np.sum(sorted_samples < x, axis=-1)] for i,x in enumerate(xs)])
```
